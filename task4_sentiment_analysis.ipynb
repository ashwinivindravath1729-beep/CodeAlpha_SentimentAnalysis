```python
# Step 1: Import Libraries
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("✅ Libraries imported!")
```

```python
# Step 2: Create Sample Review Dataset
reviews = [
    "This book is absolutely amazing and wonderful!",
    "Terrible book, waste of money and time.",
    "It was okay, nothing special about it.",
    "Loved every page, couldn't put it down!",
    "Very boring and disappointing book.",
    "Average read, some parts were good.",
    "Fantastic writing, highly recommend this!",
    "Not worth buying, very bad quality.",
    "Decent book, enjoyed reading it.",
    "Outstanding! Best book I have ever read!",
    "Horrible experience, very poorly written.",
    "Pretty good book, liked the story.",
    "Excellent content, very informative!",
    "Did not enjoy it at all, very dull.",
    "Neutral feelings, it was just fine."
]

df_reviews = pd.DataFrame({'Review': reviews})
print("✅ Dataset Created!")
print(f"Total Reviews: {len(df_reviews)}")
df_reviews.head()
```

```python
# Step 3: Analyze Sentiment using VADER
analyzer = SentimentIntensityAnalyzer()

def get_sentiment(review):
    score = analyzer.polarity_scores(review)
    if score['compound'] >= 0.05:
        return 'Positive'
    elif score['compound'] <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df_reviews['Sentiment'] = df_reviews['Review'].apply(get_sentiment)

print("✅ Sentiment Analysis Done!")
print(df_reviews)
```

```python
# Step 4: Sentiment Distribution Chart
plt.figure(figsize=(8, 6))
sns.countplot(x='Sentiment', data=df_reviews, 
              palette={'Positive':'#2ecc71', 'Negative':'#e74c3c', 'Neutral':'#3498db'})
plt.title('📊 Sentiment Distribution of Book Reviews', fontsize=16)
plt.xlabel('Sentiment', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.tight_layout()
plt.savefig('chart_sentiment_distribution.png')
plt.show()
print("✅ Chart saved!")
```

```python
# Step 5: Final Summary
print("=" * 40)
print("  SENTIMENT ANALYSIS SUMMARY")
print("=" * 40)
print(f"Total Reviews Analyzed : {len(df_reviews)}")
print(f"Positive Reviews       : {(df_reviews['Sentiment'] == 'Positive').sum()}")
print(f"Negative Reviews       : {(df_reviews['Sentiment'] == 'Negative').sum()}")
print(f"Neutral Reviews        : {(df_reviews['Sentiment'] == 'Neutral').sum()}")
print("=" * 40)
print("✅ Task 4 - Sentiment Analysis Complete!")

# Save to CSV
df_reviews.to_csv('sentiment_results.csv', index=False)
print("✅ Results saved as sentiment_results.csv!")
```
